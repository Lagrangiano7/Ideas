1. Walk: a sequence where you start in one vertex and reach another only through adjacent nodes (if the initial and final nodes coincide, we say the walk is "closed")
2. Trail: a walk that does not repeat edges
3. Path: a walk that does not repeat nodes (except closed paths, which are called "cycles"). path $\Longrightarrow$ trail
4. Eulerian trail: a closed trail that includes all edges of the graph
    - If the graph allows for at least one eulerian path, it's called "eulerian graph"
    - **Theorem:** $G$ is eulerian $\Longleftrightarrow$ all its nodes have even degree (and $G$ is connected)
5. Hamiltonian path: a closed path that does not repeat nodes
    - If the graph allows for at least one Hamiltonian path, it's called "hamiltonian graph"
    - **Ore's theorem:** for every pair of non-adjacent nodes $(v, w)$ we have $deg(v)+deg(w)\geq n \Longrightarrow$ $G$ is hamiltonian
    - **Dirac's corollary:** every node has $deg(v)\geq n/2 \Longrightarrow$ $G$ is hamiltonian
