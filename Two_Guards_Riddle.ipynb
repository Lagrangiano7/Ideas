**Resolution to the two guards riddle**

**Setting:** there are two guards, each guarding one door. One of the doors leads to Heaven, the other to Hell. One of the guards always tells the truth, the other always lies. You can only ask one question. How can you get to Heaven?

**Proposition:** you must approach either of the guards and ask "if I ask the other guard whether Heaven is behind you, will he say yes?" If the say "no", then Heaven is behind them. If they say yes, Heaven is in the other door

**Proof:** although there are 2 combinations: either Truth is guariding Heaven or he is guarding hell, since we cannot know who we are asking, we must examine all 4 possibilities:

Case 1: Truth guarding Heaven + ask Truth --> No

Case 2: Lie guarding Heaven + ask Lie --> No

Case 3: Truth guarding hell + ask Truth --> Yes

Case 4: Lie guarding hell + ask Lie --> Yes

Therefore, only standing in front of Heaven makes you say "No" and only standing in front of hell makes you say "Yes". If you get a "No", enter that door. If you get a "Yes", choose the other door$_\blacksquare$
