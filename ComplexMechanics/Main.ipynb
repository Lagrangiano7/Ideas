# Complex number methods for Classical Mechanics

**Problem statement:** when working on mechanis on the plane, it is more convenient to adopt coordinates adapted to said plane, for instance, polar coordinates. However, these entail a few inconveniences, if not difficulties:

1. The coordinate system ($\vec e_\rho, \vec e_\theta$) is no longer constant and therefore it must be derived twice in Newton's equations
2. Differential operators such as the Laplacian acquire more complicated forms than they would have in Cartesian coordinates

**Objective:** develop a formulation of orbital mechanics in employing complex numbers

**Advantages:**

1. Compression:
    - Of coordinates: instead of working with a two-component vector, we can compress both degrees of freedom (radial and angular) into a single complex number: $z=\rho e^{i\theta}$
    - Of equations: since we only have one mechanical variable (which contains two degrees of freedom at the same time, this is impossible with real numbers unless you resort to vectors, but in complex numbers it is actually the natural thing to have), we no longer need to write one equation for each component but one for all
2. Analytic continuation?:
    -  If the system is linear, we can solve it with a Green function and study its analytic continuation to formulate a dispersion relation: the "y" component could be known in terms of the "x" component, I think
    -  Bound states could manifest as poles of the Green function (this is harder because most potentials allowing bound states are not linear)



# 1. Formulation: complex Newton

Let us have a given orbital mechanical system such that its movement is confined in a plane. Let $\mathcal O$ be the origin selected in that plane and $\vec x$ be the position of the particle from said origin. Newton's equations then let us determine its dynamics with prior knowledge of the force:

$$m\ddot{\vec x}(t)=\vec F(\vec x(t))$$

**Proposition:** let us compress the two degrees of freedom in a single, complex variable: $z=\rho e^{i\theta}$. Then, we have the following properties:

1. Cartesian to polar is natural, it is neatly encoded in real / imag. parts of $z$: $x=Re z=\rho\cos\theta; \ y = Im z=\rho\sin\theta$, and $z=x+iy$
2. Same derivatives: $\vec x^{(n)} \leftrightarrow z^{(n)}$
3. Same modulus: $|z|^2=|\vec x|^2=\rho^2$
4. Same modulus for derivatives: $|z^{(n)}|^2=|\vec x^{(n)}|^2$

**Proofs:**

1. Trivial
2. From 1, we have: $z = Re z + i Im z = x + i y$. Therefore, $z^{(n)}=x^{(n)}+iy^{(n)}\leftrightarrow (x^{(n)}, y^{(n)})=\vec x^{(n)}$
3. Immediate
4. Reasoning as in 2: $|z^{(n)}|^2=(x^{(n)})^2+(y^{(n)})^2=|\vec x^{(n)}|^2$

**Def.:** if we understand the complex $z$-plane as the actual plane in space where the particle is moving, then its trajectory will now be $z(t): \mathbb R\to \mathbb C$ tq. $z$ is such that its coordinates fulfill the usual Newton's equations

**Newton's 2nd law for $z$:** $$\boxed{m\ddot z(t)=\mathcal F(z(t))}$$

where $\mathcal F = F_x+iF_y$

**Proof:** Newton's equations in cartesian coordinates are $\begin{cases}m\ddot x = F_x \\ m\ddot y = F_y\end{cases}$. Now, let us multiply the second one by the imaginary unit and add them up, obtaining the complex form of Newton's second law. It is now a complex equation, and therefore the real and imaginary parts of said equation must be fulfilled separately, so both equations are equivalent, we have simply compressed two equations into one in the same fashion as vector notation does, with the difference that we can now understanding it as a single equation for $z(t)$

# 2. Examples

## 2.1 Harmonic oscillator

Newton's 2nd law reads: $m\ddot z = -kz$, so we may solve it in the same way as we would solve this ODE in cartesian coordinates. Let us rewrite the equation as $\ddot z + \omega^2 z = 0$, so the solution is $z(t)=Ae^{i\omega t}+Be^{-i\omega t}$, where $A,B\in \mathbb C$ now, because the ODE is complex, so the solution space is still 2-dimensional but the vector space is now defined on the $\mathbb C$ field, not $\mathbb R$. Therefore, we have:

$$x=(Re A+Re B)\cos(\omega t)+(Im B-Im A)\sin(\omega t)$$
$$y=(Im A+Im B)\cos(\omega t)+(Re A-ReB)\sin(\omega t)$$

Where, naturally, all the coefficients are independent, as we would have obtained from solving the cartesian ODE (we get a decoupled, 2nd-order system, so 2 constants for each coordinate makes 4 coefficients)

**Observation:** we have halved the number of coefficients we need, but the trick is these are promoted to complex cofficients, so the number of real coefficients is always the same

We may now further compactify the notation by defining: $C=A+B, D=A-B$, to make it more evident that only two complex coefficients are needed, but this still translates to 4 real coefficients:

$$x=Re C\cos(\omega t)-Im D\sin(\omega t)$$
$$y=Im C\cos(\omega t)+Re D\sin(\omega t)$$

Now let's associate these coefficients with the initial kinematic conditions:

- $Re C = x_0$
- $Im C = y_0$
- $Re D = v_{y,0}/\omega$
- $Im D = -v_{x,0}/\omega$

So we can associate $C\leftrightarrow \vec x_0$, but this association is blurred for $D\cancel \leftrightarrow \vec v_0$

## 2.2 Keplerian potential

Another of the more studied potentials in polar coordinates is the Keplerian potential, which induces a central force: $\vec F=-\frac{\alpha}{\rho^2}\vec e_\rho$. Therefore, this gives place to the complex force: $\mathcal F = -\frac{\alpha}{\rho^2}(\cos\theta+i\sin\theta)=-\frac{\alpha}{|z|^3}z$

Hence, the problem we need to solve is:

$$\ddot z+\frac{\alpha}{|z|^3}z=0$$

**Route 1:** expand $z=\rho e^{i\theta}$:

$\dot z = \dot \rho e^{i\theta} + i\rho\dot\theta e^{i\theta}$

$\ddot z = [\ddot \rho + i\rho\ddot \theta +2i\dot \rho\dot \theta - \rho\dot  \theta^2]e^{i\theta}$

Introducing this expression in Newton's law:

$$\ddot \rho + i\rho\ddot \theta +2i\dot \rho\dot \theta - \rho\dot  \theta^2+\frac{\alpha}{\rho^2}=0$$

This neatly decouples into 2 expressions:



$$\begin{cases}\ddot \rho -\rho \dot\theta^2+\frac{\alpha}{\rho^2}=0 \\ \rho\ddot \theta+2\dot \rho\dot \theta=0\end{cases}$$

**Observation:** the second equation is but the conservation of angular momentum, which can be written as $\frac{d}{dt}(\rho^2 \dot \theta)=0$

This can be readily solved using the usual methods (conserved quantity, etc.)

**Route 2: complex change of variable** (Levi-Civita + Sundman)

Step 1: analytic change of variable: $z=\omega^2$

Step 2: change of parameter: $dt = |z|ds=|\omega|^2ds$

We now calculate $\ddot z$:

$\frac{d}{dt}=\frac{1}{\rho}\frac{d}{ds}=\frac{1}{|\omega|^2}\frac{d}{ds}$

$\dot z = \frac{2\omega\omega'}{|\omega|^2}=2\frac{\omega'}{\bar\omega}$


$\ddot z = 2\frac{1}{|\omega|^2}\frac{\omega''}{\bar\omega}-2\frac{\omega'}{|\omega|^2\bar\omega^{2}}\bar\omega'=\frac{2}{|\omega|^2\bar\omega^2}(\omega''\bar\omega-\omega'\bar \omega')=\frac{2}{\omega\bar\omega^3}(\omega''\bar\omega-|\omega'|^2)$

Now, for the potential:

$\frac{\alpha}{|z|^3}z=\frac{\alpha}{|\omega|^6}\omega^2=\frac{\alpha}{\omega^3\bar\omega^3}\omega^2=\frac{\alpha}{\omega\bar\omega^3}$

Now we substitute in the complex 2nd law:

$\frac{2}{\omega\bar\omega^3}(\omega''\bar\omega-|\omega'|^2)+\frac{\alpha}{\omega\bar\omega^3}=0$

Removing the denominator:

$2(\omega''\bar\omega-|\omega'|^2)+\alpha=0$

Step 3: further simplify the equation by using a conserved quantity: the energy, $E$

$E=\frac{1}{2}|\dot z|^2-\frac{\alpha}{|z|}=cte.$

In terms of our new variable:

$E=\frac{1}{2}|2\frac{\omega'}{\bar\omega}|^2-\frac{\alpha}{\omega\bar\omega}=2\frac{|\omega'|^2}{\omega\bar\omega}-\frac{\alpha}{\omega\bar\omega}$

Notice this bit resembles the second and third terms in the equation of motion:

$\alpha-2|\omega'|^2=-E\omega\bar\omega$

Substituting in the equation of motion:

$2\omega''\bar\omega -E\omega\bar\omega=0$

Finally getting to: $\boxed{\omega''-\frac{E}{2}\omega=0}$

**Note:** if the trajectory is a bound state, $E<0$, then we have an harmonic oscillator in this new variable

**Note:** in order to solve the problem, we now:

1. Solve the ODE for $\omega(s)$
2. Solve $dt = |\omega(s)|^2ds$ and invert for $s(t)$
3. Do: $z(t)=\omega^2(s(t))$

**Generalization:** the generalization of this phenomenon is Arnold-Bohlin duality. In particular, the Keplerian problem is dual with the harmonic oscillator, as we have seen

# 3. Complex Lagrangian dynamics

**Proposition:** for central potentials, $V(r)$, we may express the Lagrangian in terms of $z$ as:

$$L=\frac{1}{2}m|\dot z|^2-V(|z|)$$

**Proposition:** for arbitrary potentials, $V(\vec x)$, we can do $V(z)$. An anisotropy (e.g. appearance of $x$ but not of $y$) may be expressed in terms of $Re z$ and $Im z$, which in turn can be expressed in terms of $z\pm \bar z$

**Proposition:** the Euler-Lagrange equations can be expressed as:

$$\frac{\partial L}{\partial z}-\frac{d}{dt}\left(\frac{\partial L}{\partial z}\right)=0$$

$$\frac{\partial L}{\partial \bar z}-\frac{d}{dt}\left(\frac{\partial L}{\partial \bar z}\right)=0$$

**Note:** this claim comes directly from developing, for instance, the first expression: $\frac{\partial L}{\partial x}-i\frac{\partial L}{\partial y}-\frac{d}{dt}\frac{\partial L}{\partial x}+i\frac{d}{dt}\frac{\partial L}{\partial y}=0$. I include the second one for completeness, I'm not sure if it is redundant, but some anisotropies may be need the second equation

## 3.1 Example: harmonic oscillator

$$L=\frac{1}{2}m\dot z\dot{\bar z}-\frac{1}{2}m\omega^2 z\bar z$$

Now we apply E-L:

$-\frac{1}{2}m\omega^2 z-\frac{d}{dt}\left(\frac{1}{2}m\dot z\right)=0$ --> $\boxed{\ddot z+\omega^2z=0}$

## 3.2 Example: Keplerian potential

$L=\frac{1}{2}m\dot z\dot{\bar z}+\frac{\alpha}{|z|}=\frac{1}{2}m\dot z\dot{\bar z}+\frac{\alpha}{\sqrt{z\bar z}}$

Now we apply E-L:

$\frac{\alpha}{\sqrt z}(-1/2)\bar z^{-3/2}-\frac{d}{dt}\left(\frac{1}{2}m\dot z\right)=0$

Note: let us absorb the mass inside the parameter

$\ddot z+\frac{\alpha}{\bar z\sqrt {z\bar z}}=0$

Finally, let us multiply and divide by $z$, and rewrite $\sqrt{z\bar z}=|z|$ to get the complex equations of motion:

$$\boxed{\ddot z + \frac{\alpha}{|z|^3}z=0}$$
