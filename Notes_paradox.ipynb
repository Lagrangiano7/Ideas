**Formulation:** Suppose you have two notes in front of you. Note A reads "this statement is true" and note B reads "this statement is false". Your job is to determine whether these statements are really true or false by assigning them either a green tick or a red cross.

**Proposition:** it is impossible to determine whether note B is true or false because its veracity is not well defined

**Proof:** let's say we decide to claim note B is true. Therefore, the statement "this statement is false" is true, and we should have assigned a cross instead. Let us correct our judgement and claim note B is false instead. Now, the statement "this statement is false" is false, and hence the statement must be true, and we should assign a tick, thus completing the loop and closing the paradox. Therefore, neither true nor false work, and hence we can say the veracity of this statement is not well defined, as we cannot assign it a boolean value $_\blacksquare$

**Proposition:** note A does admit judgement, but we cannot know which value we should assign

**Proof:** let us claim note A is true. Then, we would have that "this statement is true" is true, and that is valid. Now let us claim it is false. Then, we would have "this statement is true" is false, and hence the note would simply be telling a lie, but that is still valid, the note is simply lying, so it is false. Therefore, unlike note B, to which we could not assign a true or false value, to note A we can, but we cannot determine which it is, as we cannot tell whether it is really telling the truth or lying $_\blacksquare$

**Comments:** the problem presented by each note is different: note B does not have a well defined true or false value, so it does not even make sense to talk about its truthfulness, as this is not defined. Since logic cannot be used to determine its veracity, this problem lies outside of logic. On the other hand, note A does admit a true or false value, but we need more information than just the statement itself to determine its truthfulness. Therefore, the exercise with note A is logical, but we lack information
