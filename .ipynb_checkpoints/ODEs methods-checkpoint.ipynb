# Taylor polynomial for order 1 problem

- Let us consider the Cauchy problem:

$$\begin{cases}\frac{dx}{dt}=f(x, t) \\ x(t_0)=x_0\end{cases}$$

We want to solve this ODE. In order to do that, admitting $x(t)$ is analytic, then we would know the solution if only we knew all the coefficients to its Taylor series. But we do! Since we know exactly the value of the function at some point and since we know the value of the first derivative at all points (that is precisely what the ODE is telling us), then we can get the 0th and 1st order terms in the Taylor series. Now, for higher order derivatives, we simply derive the former order derivative and evaluate. We can truncate the series when we see fit and thus we have an approximate solution.

$$x(t)=\displaystyle\sum_{n=0}^\infty \frac{x^{(n)}}{n!}(t-t_0)^n = x(t_0)+x'(t_0)(t-t_0)+\ldots$$

The 0th order term we know, since it's the datum: $x(t_0)=x_0$. The 1st order termm we also know, since it is given to us directly evaluating in the derivative: $x'(t_0)=f(x(t_0), t_0)=f(x_0, t_0)$. Now let us find the second derivative, taking the time derivative of the first one and applying the chain rule:

$$x''(t)=\frac{\partial f}{\partial x}\dot x+\frac{\partial f}{\partial t}$$

And since we know the value of the first order derivative:

$$\boxed{x''(t)=\frac{\partial f}{\partial x}f(x, t)+\frac{\partial f}{\partial t}}$$


- We can find the next derivative in an iterative fashion, just doing:

$$x''(t)=g(x, t) \Longrightarrow x^{(3)}=\frac{d}{dt}x''=\frac{\partial g}{\partial x}f+\frac{\partial g}{\partial t}$$

- In general:

$$\boxed{x^{(n)}(t)= \frac{\partial g_{n-1}}{\partial x}f(x, t)+\frac{\partial g_{n-1}}{\partial t}}$$

- Once we have the set of the first $N+1$ derivatives (counting 0th order), then we can do:

$$\boxed{x(t)\approx \displaystyle\sum_{n=0}^{N}\frac{x^{(n)}(t_0)}{n!}(t-t_0)^n}$$

**Advantage:** after the symbolic calculus, this is just evaluation. No need to manipulate numbers any further. Plus, it only takes $N-1$ iterations (I have $N+1$ coefficients, but I only need to perform calculations to obtain from order 2 to order N, so I don't count the two first, since they can be directly evaluated)

**Advantage:** its exactitude does not depend on the number of points considered, since it is exact up to truncation. Therefore, what introduces the error is the truncation, not the approximation of derivatives, since this is performed previously symbolically

## Practical application

- Let us solve the Cauchy problem:

$$\begin{cases}\frac{dx}{dt}=-x+\cos(t) \\ x(0)=1\end{cases}$$

- We use Runge-Kutta 4 to solve the problem in parallel as a "control" method. We use 100

```python
# I want to solve the first order Cauchy problem:


"""
dx/dt = f(x, t)
x(t0)=x0

"""

# Approach (admitting x analytic on t): if we knew all the coefficients of the Taylor series of x(t) we would be able to
# reconstruct the whole function for all of its domain. Nonetheless, we can truncate the series to get an approximate
# solution around some point (for instance, t=0). We can figure out all the Taylor coefficients of the solution
# using the ODE

from sympy import *
from math import factorial
import numpy as np
import matplotlib.pyplot as plt

n = 6 # Truncate at 6 (included)

x = Symbol("x")
t = Symbol("t")

f1 = -x+cos(t)
f = -x+cos(t)
func = lambdify((x, t), f, "math")

t0=0
x0=1

coefs = [x0, func(x0, t0)] # x0, x1

for i in range(2, n+1):
    f = diff(f, x)*f1+diff(f, t)
    func = lambdify((x, t), f, "math")
    coefs.append(func(x0, t0)/factorial(i))

print(coefs)

############ RESUELVO Y COMPARO ############

f = lambda x, t: -x+np.cos(t)

tp = np.linspace(0,3,1000)
h = 3/1000
x=1
xp=[]

for t in tp:
    xp.append(x)
    k1 = h*f(x, t)
    k2 = h*f(x+k1/2, t+h/2)
    k3 = h*f(x+k2/2, t+h/2)
    k4 = h*f(x+k3, t+h)

    x+=1/6*(k1+2*k2+2*k3+k4)

plt.plot(tp, xp, "r", label="RK4")
plt.plot(tp, np.polyval(coefs[::-1], tp), label="Polynomial (n=6)")
plt.legend()
plt.show()
```

**Observation:** 5 iterations for my method, 1000 iterations for RK4
