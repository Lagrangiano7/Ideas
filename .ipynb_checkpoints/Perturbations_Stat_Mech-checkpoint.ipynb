**Context:** let us say we have a Hamiltonian $H$ for a many-body, classical system. We would like to obtain the macroscopic quantities of this system employing Boltzmann statistics (i.e. the canonical ensemble), but let us say the system is not solvable. Also, let us say the system is composed of two parts: a free term and an interaction term: $H=H_0+V$, where $H_0$ is just the kinetic term of the system. If $V<<H_0$, then it would be more natural to denote $V=\Delta H$ and call it a perturbation to the free term. However, let us keep it general, at least for now

**Objective:** as is the case in any canonical ensemble problem, we would like to compute the partition function of the system, as from the knowledge of $Z$ we may derive the macroscopic properties

**Difficulty:** since we can't solve the dynamics of the system analytically, we can't obtain $Z$

**Proposed solution:** we do a perturbative expansion in $\beta$

**Procedure:** for the full Hamiltonian, we have:

$$Z=\int d\Gamma e^{-\beta H}=\int d\Gamma e^{-\beta H_0}e^{-\beta V}$$

Now, we expand the exponential, obtaining:

$$Z=\displaystyle\sum_{n=0}^\infty \frac{(-1)^n}{n!}\beta^n \int d\Gamma \ V^n e^{-\beta H_0}$$

where the integrals now may be tractable, order by order

**Important:** this is not strictly a perturbative expansion in $\beta$, seeing as the integral also depends on this parameter. For instance, for the ideal gas with a $p^4$ ultrarelativistic correction, the integral yields terms such as $1/\beta$ to some power which make the $\beta <<1$ expansion completely useless and instead yield a good result for $\beta >> 1$ (i.e. low temperatures).

Since the zeroth-order term is known (just the free system, which by hypothesis is solvable), in practice we may factor it out:

$Z=Z_0+\displaystyle\sum_{n=1}^\infty \frac{(-1)^n}{n!}\beta^n \int d\Gamma \ V^n e^{-\beta H_0}$

More compactly:

$Z=Z_0+\displaystyle\sum_{n=1}^\infty \frac{(-1)^n}{n!}\beta^n G_n(\beta)$, with $G_n(\beta)= \int d\Gamma \ V^n e^{-\beta H_0}$

**Proposed (truly perturbative) approach:** if the integral allows, we may a second expansion, now in $G_n(\beta)$, to obtain something truly perturbative in $\beta$:

$$Z=\displaystyle\sum_{n,m=0}^\infty \frac{(-1)^n}{n!m!}G_n^{(m)}|_{\beta=0}\beta^{n+m}$$

If we have $\beta << 1$, we may stay at firs-order in $\beta$, so the only possible combinations in this double-series expansion are $(n,m)=(0,0), (1,0), (0,1)$, and we get:

$$Z\approx G_0|_{\beta=0}-G_1|_{\beta=0}\beta+G_0'|_{\beta=0}\beta$$

Naming the constants: $A_i=G_i|_{\beta=0}$ and $B_i=G'_{i}|_{\beta=0}$, we have_

$$\boxed{Z\approx A_0+(B_0-A_1)\beta}$$
